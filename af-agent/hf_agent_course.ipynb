```python
from dotenv import load_dotenv
import os
```

```python
_ = load_dotenv()
```

```python
# Pass the token to InferenceClient/Model
token = token=os.getenv("HF_TOKEN")
model = InferenceClientModel(token=token)
#print (token)

# Create agent without token parameter
agent = CodeAgent(
    tools=[], 
    model=model
)
```

```python
# Run the agent with a task
result = agent.run("Calculate the sum of numbers from 1 to 10")
print(result)
```

```python
from smolagents import DuckDuckGoSearchTool, VisitWebpageTool


# Test the tool directly first
print("=== Testing DuckDuckGo directly ===")
search_tool = DuckDuckGoSearchTool()

try:
    results = search_tool("party music recommendations")
    print(f"Direct results: {results}")
except Exception as e:
    print(f"Error: {e}")
    import traceback
    traceback.print_exc()


agent = CodeAgent(
    tools=[DuckDuckGoSearchTool(),
          VisitWebpageTool() ], 
    model=model)

agent.run("Search for the best music recommendations for a party at the Wayne's mansion.")
```

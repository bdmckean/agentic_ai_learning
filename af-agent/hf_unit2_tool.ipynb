```python
from smolagents import ToolCallingAgent, WebSearchTool, InferenceClientModel
from dotenv import load_dotenv
import os

load_dotenv()

# Use a model specifically designed for tool calling
model = InferenceClientModel(
    model="Qwen/Qwen2.5-Coder-32B-Instruct",  # This model has great tool support
    token=os.getenv("HF_TOKEN")
)

# Initialize the search tool
search_tool = WebSearchTool()

# Create agent
agent = ToolCallingAgent(
    tools=[search_tool], 
    model=model
)

# Run the task
try:
    response = agent.run("Search for the best music recommendations for a party at the Wayne's mansion.")
    print(f"\n🎉 Result:\n{response}")
except Exception as e:
    print(f"\n❌ Error occurred: {e}")
```

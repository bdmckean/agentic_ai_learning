```python
# Run the agent - Simplified version without Hub loading
from smolagents import CodeAgent, InferenceClientModel
import os

prompt = "Give me the best playlist for a party at Wayne's mansion. The party idea is a 'villain masquerade' theme"

print("Creating agent...")

# Create agent directly without Hub to avoid Jupyter notebook issues
model = InferenceClientModel(token=os.getenv("HF_TOKEN"))
agent = CodeAgent(tools=[], model=model)

print("Running agent...")
result = agent.run(prompt)

print(f"\n✓ Agent completed successfully")
print(f"\nResult preview:\n{str(result)[:500]}...")
```

```python
# Send trace to Langfuse (run this AFTER cell 8 completes)
from langfuse import Langfuse

langfuse = Langfuse(
    public_key=os.getenv('LANGFUSE_PUBLIC_KEY'),
    secret_key=os.getenv('LANGFUSE_SECRET_KEY'),
    host=os.getenv('LANGFUSE_HOST')
)

# Use the prompt and result from cell 8
prompt = "Give me the best playlist for a party at Wayne's mansion. The party idea is a 'villain masquerade' theme"

span = langfuse.start_span(
    name="Alfred Agent Run",
    input={"prompt": prompt}
)

# Update span with output, then end it
span.update(output={"result": str(result)[:2000]})
span.end()
langfuse.flush()

print("✓ Trace sent to Langfuse dashboard")
```
